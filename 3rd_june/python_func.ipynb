```python
def volume(l=1 ,b=1,h=1):
    return l*b*h

print(volume()) #default values
print(volume(1,2,3)) #custom values
print(volume(l=7, b=6, h=3)) #custom values with keyword arguments
```

## Variable length argument

```python
def f(*x):
    return x

print(f(1,2,3))
```

```python
#right a fun to find the avg of n no.
def avg(*x):
    return sum(x)/len(x)
print(avg(1,2,3,4,5))
   

```

```python
import matplotlib.pyplot as plt
from collections import Counter

# Show the plot for the frequency of the words in the name
name = "rdfwjobbrgbrebnoenrorvron"

letter_counts = Counter(name)

letters = list(letter_counts.keys())
frequencies = list(letter_counts.values())

print(letter_counts.keys())
print(letter_counts.values())
plt.bar(letters, frequencies)

plt.xlabel("Letters")
plt.ylabel("Frequency")

plt.show()
```
