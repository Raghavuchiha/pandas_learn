```python
# translate english to german using dictionary

translate = {}
translate["good"] = "gut"
translate["coffee"] = "kaffee"
translate["milk"] = "milch"
translate["please"] = "bitte"
translate["thank you"] = "danke"
translate["bye"] = "tschüss"
translate["hello"] = "hallo"
translate["world"] = "welt"
translate["python"] = "python"
translate["father"] = "vater"

def translate_to_german(word):
   if word not in translate:
         return "word not found"
   return translate[word]

print(translate_to_german("mother"))  # Output: "tschüss"
```

```python
emojis = {
    "happy": "😊",
    "sad": "😔",
    "love": "😍",
    "frustrated": "😣",
    "serious": "🫨"
}

sentence = "I am happy"

def replace_mood(sentence):
    words = sentence.split()
    for i in range(len(words)):
        if words[i] in emojis:
            words[i] = emojis[words[i]]
    return " ".join(words)

print(replace_mood(sentence))
```

```python
emojis = {
    "happy": "😊",
    "sad": "😔",
    "love": "😍",
    "frustrated": "😣",
    "serious": "🫨"
}

sentence = "I am happy"
words = sentence.split()
result = ""

for i in words:
    if i in emojis:
        result += emojis[i] + " "
    else:
        result += i + " "

print(result)
```
